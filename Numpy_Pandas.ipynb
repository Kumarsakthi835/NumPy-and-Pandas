```python
#1D NumPy Array
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 Temperatures:", temperatures_w1)
```

```python
#Inspection and Properties
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#Array Operations

#Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temperatures_f)

# Max, Min, Mean
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
#Array Slicing and Indexing

# First three days
print("First 3 days:", temperatures_w1[:3])

# last two days
print("Weekend:", temperatures_w1[-2:])

# Middle three days
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
#Creating Pandas Series
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("Marks Series:")
print(marks)
```

```python
#Indexing and Slicing
print("1st Rank Mark (integer index):", marks[0])
print("Top 3 Ranks (loc):")
print(marks.loc['Rank1':'Rank3'])
print("3rd Rank Mark (iloc):", marks.iloc[2])
print("Marks greater than 90:")
print(marks[marks > 90])
```

```python
#Manipulating Series

# Modify 1st rank mark to 100
marks['Rank1'] = 100
print("After modifying Rank1:")
print(marks)

# Remove last rank
marks = marks.drop('Rank5')
print("After removing Rank5:")
print(marks)

# Compute CGPA
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
#Creating Pandas DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North',
               'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python
#Data Exploration
# Head and Tail
print("Head:")
print(transactions.head())
print("Tail:")
print(transactions.tail())
# Shape
print("Shape:", transactions.shape)
# Column names
print("Columns:", transactions.columns.tolist())
# Data types
print("Data Types:")
print(transactions.dtypes)
# ProductCategory and Amount columns
print("ProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])
# Last 3 columns
print("Last 3 columns:")
print(transactions.iloc[:, -3:])

# Filter North region and Amount > 200
print("North Region with Amount > 200:")
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

# Value counts for ProductCategory
print("Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# Unique values in Region
print("Unique Regions:", transactions['Region'].unique())

# Group by Region - mean amount
print("Mean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#DataFrame Manipulation
# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("After modifying TransactionID 102:")
print(transactions)
# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10
print("After adding Discount column:")
print(transactions)
# Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
print("After removing TransactionID 109:")
print(transactions)
# Delete Discount column
transactions = transactions.drop(columns=['Discount'])
print("After deleting Discount column:")
print(transactions)
```
